1-mashq — Murakkab funksiyani bosqichma-bosqich hisoblash


```python
#1) Funksiya: y = (5x + 2)³

#2) Ichki funksiya g(x) va tashqi funksiya f(u) ni ayting (Dars 26, 7-slayddagi shablon asosida).

def g(x):
  return 5*x + 2

def f(x):
  return g(x)**3


# 3) x = 1, 2, 3 uchun avval u = g(x) ni, keyin y = f(u) ni bosqichma-bosqich hisoblang.

for x in [1, 2, 3]:
  print(f"u = g({x}) => {g(x)}")

for x in [1, 2, 3]:
  print(f"y = f({x}) => {f(x)}")
```

2-mashq — Oddiy chain rule bilan hosila topish


```python
#1) Funksiya: y = (x − 4)²


#2) Ichki va tashqi funksiyalarni ajratib, chain rule bilan dy/dx formulasini QO'LDA toping (Dars 26, 11-slayddagi formula: dy/dx = dy/du · du/dx).

dydu = 1
dudx = 2*(x - 4)

#3) x = 0, 4, 6 nuqtalarida dy/dx qiymatlarini hisoblang.
# dy/dx = dy/du · du/dx =>:
# 1 * 2*(0-4) = -8
# 1 * 2*(4-4) = 0
# 1 * 2*(6-4) = 4


#4) sp.diff bilan tekshiring.

import sympy as sp


x = sp.symbols('x')

f = (x - 4)**2

hosila_sym = sp.diff(f, x)

for x_val in [0, 4, 6]:
  result = hosila_sym.subs(x, x_val)
  print(f"x = {x_val}, dy/dx = {result}")
```

3-mashq — Ikki hadli ichki funksiya bilan chain rule


```python
#1) Funksiya: y = (x² − 2x)³

#2) u = g(x) = x² − 2x va f(u) = u³ ekanini aniqlab, du/dx va dy/du ni toping.

u = sp.symbols('u')
g = x**2 - 2*x
f = u**3

dudx = sp.diff(g, x)
dydu = sp.diff(f, u)

print(f"du/dx = {dudx}")
print(f"dy/du = {dydu}")

# 3) Chain rule bilan dy/dx formulasini QO'LDA chiqaring.

dydx = 3*(x**2-2*x)**2 * (2*x-2)
print(f"dy/dx = {dydx}")


# 4) x = 0, 1, 3 nuqtalarida qiymatlarini hisoblang va sp.diff bilan tekshiring.

# x = 0:
#dydx = 3(0^2-2*0)^2 * (2*0-2)= 0

# x = 1:
#dydx = 3(1^2-2*1)^2 * (2*1-2)= 0

# x = 3:
#dydx = 3(3^2-2*3)^2 * (2*3-2)= 108

for x_val in [0, 1, 3]:
  result = dydx.subs(x, x_val)
  print(f"x = {x_val}, dy/dx = {result}")
```

5-mashq — Mini-neyronni to'liq o'qitish (backprop)

```python
#1) h = w1·x + b1, a = h², loss = (a − target)² formulalarini kod bilan hisoblang.

x, w1, b1, target = 3, 0.8, 0.5, 15
h = w1 * x + b1
a = h ** 2
loss = (a - target) ** 2
print(f"h: {h:.2f}")
print(f"a: {a:.2f}")
print(f"loss: {loss:.2f}")

#2) Chain rule bilan dLoss/dw1 = dLoss/da · da/dh · dh/dw1 ni hisoblang.

dLoss_da = 2 * (a - target)
da_dh = 2 * h
dh_dw1 = x
dLoss_dw1 = dLoss_da * da_dh * dh_dw1

# 3) Xuddi shunday dLoss/db1 = dLoss/da · da/dh · dh/db1 ni hisoblang (Dars 26, mashq 4 shablonidan foydalaning).

dh_db1 = 1
dLoss_db1 = dLoss_da * da_dh * dh_db1


# 4) Ikkala gradient ham manfiy yoki musbat ekanini tekshirib, w1 va b1 ni qaysi tomonga o'zgartirish kerakligini ayting (gradient descent mantig'i).
print(f"Gradient of w: {dLoss_dw1:.3f}")
print(f"Gradient of w: {dLoss_db1:.3f}")
#Javob: 2 lasi ham manfiy demak w ni ham b ni ham oshirish kerak + qilish kerak, minimumdan chapga qarab ketib qolgan.
```
