🔥 Task 3 — Name'dan ma'lumot qazib ol

```python
df['Name']
```

```python
df['Title'] = df['Name'].str.extract(r'\s*, ([^\.]+)\.')
df['Title']
```

```python
df['Title'].value_counts()
```

🔥 Task 4 — Rare Titles

```python
for value in df['Title'].value_counts().index:
  if df['Title'].value_counts()[value] <= 2:
    df['Title'] = df['Title'].replace(value, 'Rare')
```

```python
df['Title'].value_counts()
```

🔥 Task 5 — Fare Per Person

```python
import numpy as np

df['fare_per_person'] = df['Fare'] / df['family_size']

df['fare_per_person'].describe()
```

```python
print(df['fare_per_person'].mean(),
df['fare_per_person'].min(),
df['fare_per_person'].max())
```

```python
df['fare_per_person'] = np.round(df['fare_per_person'], 2)
```

🔥 Task 6 — Yana bitta mustaqil Feature

```python
# df['age_group'] = pd.cut(df['Age'], bins=[0, 18, 60, 80], labels=['child','adult', 'senior'])


# df['age_group'].value_counts()

```

```python
age_groups = []

for value in df['Age']:
  if value < 18:
    age_groups.append('child')
  elif value >= 18 and value < 60:
    age_groups.append('adult')
  else:
    age_groups.append('senior')

df['age_group'] = age_groups
```

```python
df['age_group'].value_counts()
```
